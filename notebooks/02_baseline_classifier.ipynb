# 02 - Baseline Classifier (TF-IDF + Logistic Regression)

Trains and evaluates the baseline category classifier used by default in the running pipeline.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import pandas as pd
from src.config import settings
from src.classification.baseline import BaselineClassifier

df = pd.read_csv(settings.path('paths.complaints_csv'))
df['complaint_text'] = df['complaint_text'].fillna('')
df.shape
```

```python
clf = BaselineClassifier(target_name='category')
metrics = clf.fit(df['complaint_text'].tolist(), df['category'].tolist())
print('accuracy:', metrics['accuracy'])
print('macro F1:', metrics['macro_f1'])
print('weighted F1:', metrics['weighted_f1'])
```

```python
import pandas as pd
pd.DataFrame(metrics['classification_report']).T
```

```python
# Confusion matrix
import numpy as np
cm = pd.DataFrame(metrics['confusion_matrix'], index=metrics['labels'], columns=metrics['labels'])
cm
```
