# 04 - Sentiment & Emotion Analysis

Demonstrates the sentiment and emotion analyzers (`src/sentiment/analyzer.py`, `src/emotion/analyzer.py`) and their backend selection (transformer-optional, lexicon fallback).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
from src.sentiment.analyzer import get_sentiment_analyzer
from src.emotion.analyzer import get_emotion_analyzer

sent = get_sentiment_analyzer()
emo = get_emotion_analyzer()
print('Sentiment backend:', sent.backend)
print('Emotion backend:', emo.backend)
```

```python
examples = [
    "My OTP never arrives when trying to login, this is extremely frustrating!",
    "Thank you so much, my refund was processed quickly and the support was great.",
    "The order is still pending, not sure what is going on.",
]
for t in examples:
    s = sent.analyze(t)
    e = emo.analyze(t)
    print(f"{t}\n  sentiment={s['label']} ({s['confidence']:.2f})  emotion={e['label']} ({e['confidence']:.2f})\n")
```

```python
import pandas as pd
from src.config import settings
df = pd.read_csv(settings.path('paths.complaints_csv')).sample(200, random_state=42)
results = [sent.analyze(t) for t in df['complaint_text']]
pd.Series([r['label'] for r in results]).value_counts(normalize=True)
```
