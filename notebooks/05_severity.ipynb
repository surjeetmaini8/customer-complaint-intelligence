Demonstrates the transparent, multi-factor rule-based severity engine (`src/severity/predictor.py`).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
from src.severity.predictor import get_severity_service
svc = get_severity_service()
```

```python
result = svc.assess(
    category='Fraud', subcategory='unauthorized transaction', sentiment='Negative',
    emotion='Angry', amount=45000, recent_similar_count=3, resolution_status='open',
)
result
```

```python
result_low = svc.assess(
    category='Other', subcategory='feature request', sentiment='Neutral',
    emotion='Neutral', amount=0, recent_similar_count=0, resolution_status='resolved',
)
result_low
```

Note severity is **not** simply `negative sentiment -> high severity`: it blends financial impact, fraud/security risk, frequency, operational status (unresolved), sentiment, and emotion, per configurable weights in `config.yaml: severity.weights`.
